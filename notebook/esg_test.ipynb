```python
import os
import finnhub
import pandas as pd

os.environ["FINNHUB_API_KEY"] = ""
client = finnhub.Client(api_key=os.getenv("FINNHUB_API_KEY"))

tickers = ["AAPL", "AMZN", "NVDA", "MSFT", "GOOGL"]

records = []
for ticker in tickers:
    try:
        data = client.stock_esg(ticker)  
        records.append({
            "ticker":           ticker,
            "year":             data.get("year"),
            "environmentScore": data.get("environmentScore"),
            "socialScore":      data.get("socialScore"),
            "governanceScore":  data.get("governanceScore")
        })
    except Exception as e:
        print(f"Error fetching ESG for {ticker}: {e}")


df_esg = pd.DataFrame(records)
print(df_esg.to_string(index=False))
```
