```python
import requests
import io
from PyPDF2 import PdfReader

# 1) 예시 URL — 실제 공개된 Macro Scenario PDF 링크로 바꿔주세요
pdf_url = "https://www.moodys.com/rs/123-456/images/Global_Macro_Scenario_Q3_2025.pdf"

# 2) PDF 다운로드
resp = requests.get(pdf_url, stream=True)
resp.raise_for_status()

# 3) 메모리에서 PDF 읽기
reader = PdfReader(io.BytesIO(resp.content))

# 4) 페이지별 텍스트 추출
scenario_text = []
for page in reader.pages:
    text = page.extract_text()
    if text:
        scenario_text.append(text)

# 5) 결과 출력 (첫 2페이지)
print("\n\n--- Page 1 ---\n")
print(scenario_text[0][:500])
print("\n\n--- Page 2 ---\n")
print(scenario_text[1][:500])
```
