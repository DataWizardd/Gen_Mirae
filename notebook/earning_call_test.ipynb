```python
import os
import finnhub
import pandas as pd

os.environ["FINNHUB_API_KEY"] = ""
client = finnhub.Client(api_key=os.getenv("FINNHUB_API_KEY"))

# 5개 종목의 어닝콜 캘린더
records = []
for ticker in ["AAPL","AMZN","NVDA","MSFT","GOOGL"]:
    data = client.earnings_calendar(symbol=ticker, _from="2025-07-01", to="2025-12-31")
    for item in data.get("earningsCalendar", []):
        records.append({
            "ticker": ticker,
            "date": item["date"],
            "epsEstimate": item.get("epsEstimate"),
            "epsActual": item.get("epsActual")
        })
df_calendar = pd.DataFrame(records)
print(df_calendar)
```
