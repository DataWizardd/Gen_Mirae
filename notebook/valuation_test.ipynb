```python
import yfinance as yf
import pandas as pd

# PoC 용 5개 종목 리스트
tickers = ["AAPL", "AMZN", "NVDA", "MSFT", "GOOGL"]

# 수집할 벨류에이션 지표
metrics = [
    "trailingPE",
    "forwardPE",
    "priceToBook",
    "priceToSalesTrailing12Months",
    "enterpriseToRevenue",
    "enterpriseToEbitda",
    "marketCap",
    "enterpriseValue",
    "totalRevenue",
    "netIncomeToCommon",
]

# 데이터 수집
rows = []
for ticker in tickers:
    t = yf.Ticker(ticker)
    info = t.info
    row = {"ticker": ticker}
    for m in metrics:
        row[m] = info.get(m, None)
    rows.append(row)

# DataFrame 생성 및 출력
df_valuation = pd.DataFrame(rows)
print(df_valuation.to_string(index=False))
```
