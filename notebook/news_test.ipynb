```python
from newsapi import NewsApiClient
import pandas as pd
import os

api_key = os.getenv("NEWSAPI_API_KEY")
newsapi = NewsApiClient(api_key=api_key)
tickers = ["AAPL","AMZN","NVDA","MSFT","GOOGL"]

records = []
for ticker in tickers:
    articles = newsapi.get_everything(
        q=ticker,
        language='en',
        sort_by='publishedAt',
        page_size=20
    )["articles"]
    for art in articles:
        records.append({
            "ticker":      ticker,
            "publishedAt": art["publishedAt"],
            "source":      art["source"]["name"],
            "title":       art["title"],
            "description": art.get("description"),
            "content":     art.get("content"),
            "url":         art["url"]
        })
df_news = pd.DataFrame(records)
print(df_news.head())
```

```python
import os
import finnhub
import pandas as pd

os.environ["FINNHUB_API_KEY"] = ""
client = finnhub.Client(api_key=os.getenv("FINNHUB_API_KEY"))

# 5개 종목의 어닝콜 캘린더
records = []
for ticker in ["AAPL","AMZN","NVDA","MSFT","GOOGL"]:
    data = client.earnings_calendar(symbol=ticker, _from="2025-07-01", to="2025-12-31")
    for item in data.get("earningsCalendar", []):
        records.append({
            "ticker": ticker,
            "date": item["date"],
            "epsEstimate": item.get("epsEstimate"),
            "epsActual": item.get("epsActual")
        })
df_calendar = pd.DataFrame(records)
print(df_calendar)
```
