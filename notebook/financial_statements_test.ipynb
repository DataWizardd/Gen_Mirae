```python
import yfinance as yf
import pandas as pd

# PoC 용 5개 종목 리스트
tickers = ["AAPL", "AMZN", "NVDA", "MSFT", "GOOGL"]

for ticker in tickers:
    t = yf.Ticker(ticker)
    
    # 1) 연간 손익계산서
    print(f"\n=== {ticker} Annual Income Statement ===")
    print(t.financials, "\n")
    
    # 2) 연간 대차대조표
    print(f"=== {ticker} Annual Balance Sheet ===")
    print(t.balance_sheet, "\n")
    
    # 3) 연간 현금흐름표
    print(f"=== {ticker} Annual Cash Flow ===")
    print(t.cashflow, "\n")
    
    # 4) 분기 손익계산서
    print(f"=== {ticker} Quarterly Income Statement ===")
    print(t.quarterly_financials, "\n")
    
    # 5) 분기 대차대조표
    print(f"=== {ticker} Quarterly Balance Sheet ===")
    print(t.quarterly_balance_sheet, "\n")
    
    # 6) 분기 현금흐름표
    print(f"=== {ticker} Quarterly Cash Flow Statement ===")
    print(t.quarterly_cashflow, "\n")
```
